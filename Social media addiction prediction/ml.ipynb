```python
X=df.drop('Addiction_Level', axis=1)
Y=df['Addiction_Level']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
```

```python
X_train.shape, X_test.shape, Y_train.shape, Y_test.shape
```

```python
numeric_features = X.select_dtypes(include=[np.number]).columns
categorical_features = X.select_dtypes(include=['object']).columns
```

## **Applying Column transformer**

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder,StandardScaler
ct = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numeric_features),
        ('cat', OneHotEncoder(drop='first', handle_unknown='ignore'), categorical_features)
    ]
)
```

## **Importing model** 

```python
## Importing Random Forest Regressor
from sklearn.ensemble import RandomForestRegressor
```

## **Making a pipeline**

```python
from sklearn.pipeline import make_pipeline
pipe = make_pipeline(ct, RandomForestRegressor())
```

## **Traning on Random Forest Regressor**

```python
pipe.fit(X_train, Y_train)
```

```python
y_pred = pipe.predict(X_test)
```

```python
#Predection score


from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score

mse = mean_squared_error(Y_test, y_pred)
r2 = r2_score(Y_test, y_pred)

print(f"Mean Squared Error: {mse}")
print(f"R^2 Score: {r2}")
```

```python
train_score=r2_score(Y_train, pipe.predict(X_train))
test_score=r2_score(Y_test, y_pred)
print(f"Train R^2 Score: {train_score}")
print(f"Test R^2 Score: {test_score}")
```

## **Importing other Regressor models**

```python
from sklearn.linear_model import Ridge, Lasso, ElasticNet
from sklearn.ensemble import RandomForestRegressor, ExtraTreesRegressor, GradientBoostingRegressor, HistGradientBoostingRegressor
from sklearn.svm import SVR
from sklearn.neighbors import KNeighborsRegressor
```

```python
models = {
    'Ridge': Ridge(),
    'Lasso': Lasso(),
    'ElasticNet': ElasticNet(),
    'RandomForest': RandomForestRegressor(),
    'ExtraTrees': ExtraTreesRegressor(),
    'GradientBoosting': GradientBoostingRegressor(),
    'HistGradientBoosting': HistGradientBoostingRegressor(),
    'SVR': SVR(),
    'KNeighbors': KNeighborsRegressor()
}
for name, model in models.items():
    pipe = make_pipeline(ct, model)
    pipe.named_steps['columntransformer'].set_params(sparse_threshold=0)
    pipe.fit(X_train, Y_train)
    y_pred = pipe.predict(X_test)
    mse = mean_squared_error(Y_test, y_pred)
    r2 = r2_score(Y_test, y_pred)
    print(f"{name} - Mean Squared Error: {mse}, R^2 Score: {r2}")
```

## **Ploting the performance of the all models**

```python
results = []

for name, model in models.items():
    pipe = make_pipeline(ct, model)
    pipe.fit(X_train, Y_train)
    y_pred = pipe.predict(X_test)
    
    results.append({
        'Model': name,
        'MSE': mean_squared_error(Y_test, y_pred),
        'R2': r2_score(Y_test, y_pred)
    })

results_df = pd.DataFrame(results)

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

sns.barplot(data=results_df, x='Model', y='MSE', ax=axes[0], palette='viridis')
axes[0].set_title('Mean Squared Error Comparison')
axes[0].tick_params(axis='x', rotation=45)

sns.barplot(data=results_df, x='Model', y='R2', ax=axes[1], palette='coolwarm')
axes[1].set_title('R2 Score Comparison')
axes[1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

## **Adding 3 big model**

```python
from xgboost import XGBRegressor
from lightgbm import LGBMRegressor
from catboost import CatBoostRegressor
```

```python
advanced_models = {
    'XGBoost': XGBRegressor(),
    'LightGBM': LGBMRegressor(),
    'CatBoost': CatBoostRegressor(verbose=0)
}
```

```python
from sklearn.metrics import mean_squared_error, r2_score

# Train the 3 advanced models on GPU if available; otherwise fall back to CPU

gpu_models = {
    'XGBoost': XGBRegressor(
        objective='reg:squarederror',
        n_estimators=500,
        max_depth=6,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        random_state=42,
        tree_method='hist',
        device='cuda'
    ),
    'LightGBM': LGBMRegressor(
        objective='regression',
        n_estimators=500,
        learning_rate=0.05,
        max_depth=-1,
        num_leaves=31,
        random_state=42,
        device='gpu'
    ),
    'CatBoost': CatBoostRegressor(
        loss_function='RMSE',
        iterations=500,
        learning_rate=0.05,
        depth=8,
        random_seed=42,
        verbose=0,
        task_type='GPU'
    )
}

cpu_fallbacks = {
    'XGBoost': XGBRegressor(
        objective='reg:squarederror',
        n_estimators=500,
        max_depth=6,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        random_state=42,
        tree_method='hist'
    ),
    'LightGBM': LGBMRegressor(
        objective='regression',
        n_estimators=500,
        learning_rate=0.05,
        max_depth=-1,
        num_leaves=31,
        random_state=42,
        device='cpu'
    ),
    'CatBoost': CatBoostRegressor(
        loss_function='RMSE',
        iterations=500,
        learning_rate=0.05,
        depth=8,
        random_seed=42,
        verbose=0,
        task_type='CPU'
    )
}

advanced_results = []

for name, model in gpu_models.items():
    pipe = make_pipeline(ct, model)
    try:
        pipe.fit(X_train, Y_train)
        used_device = "GPU"
    except Exception as e:
        print(f"{name} GPU training failed: {e}. Falling back to CPU.")
        pipe = make_pipeline(ct, cpu_fallbacks[name])
        pipe.fit(X_train, Y_train)
        used_device = "CPU"

    y_pred = pipe.predict(X_test)
    train_pred = pipe.predict(X_train)

    advanced_results.append({
        'Model': name,
        'Used_Device': used_device,
        'Train_R2': r2_score(Y_train, train_pred),
        'Test_R2': r2_score(Y_test, y_pred),
        'Train_RMSE': np.sqrt(mean_squared_error(Y_train, train_pred)),
        'Test_RMSE': np.sqrt(mean_squared_error(Y_test, y_pred))
    })

advanced_results_df = pd.DataFrame(advanced_results)
print(advanced_results_df.sort_values('Test_R2', ascending=False))
```

```python
# Create visualizations for CatBoost predictions
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# 1. Actual vs Predicted (Test Set)
axes[0, 0].scatter(Y_test, y_pred, alpha=0.6, edgecolors='k')
axes[0, 0].plot([Y_test.min(), Y_test.max()], [Y_test.min(), Y_test.max()], 'r--', lw=2)
axes[0, 0].set_xlabel('Actual Addiction Level', fontsize=12)
axes[0, 0].set_ylabel('Predicted Addiction Level', fontsize=12)
axes[0, 0].set_title('CatBoost: Actual vs Predicted (Test Set)', fontsize=14, fontweight='bold')
axes[0, 0].grid(True, alpha=0.3)

# 2. Residuals (Test Set)
residuals = Y_test.values - y_pred
axes[0, 1].scatter(y_pred, residuals, alpha=0.6, edgecolors='k')
axes[0, 1].axhline(y=0, color='r', linestyle='--', lw=2)
axes[0, 1].set_xlabel('Predicted Addiction Level', fontsize=12)
axes[0, 1].set_ylabel('Residuals', fontsize=12)
axes[0, 1].set_title('CatBoost: Residuals Plot (Test Set)', fontsize=14, fontweight='bold')
axes[0, 1].grid(True, alpha=0.3)

# 3. Distribution of Residuals
axes[1, 0].hist(residuals, bins=30, edgecolor='black', alpha=0.7, color='skyblue')
axes[1, 0].axvline(x=0, color='r', linestyle='--', lw=2)
axes[1, 0].set_xlabel('Residuals', fontsize=12)
axes[1, 0].set_ylabel('Frequency', fontsize=12)
axes[1, 0].set_title('CatBoost: Distribution of Residuals', fontsize=14, fontweight='bold')
axes[1, 0].grid(True, alpha=0.3, axis='y')

# 4. Prediction Error
prediction_error = np.abs(residuals)
axes[1, 1].bar(range(len(prediction_error[:100])), prediction_error[:100], alpha=0.7, color='coral', edgecolor='black')
axes[1, 1].set_xlabel('Sample Index (First 100)', fontsize=12)
axes[1, 1].set_ylabel('Absolute Error', fontsize=12)
axes[1, 1].set_title('CatBoost: Absolute Prediction Error (First 100 Samples)', fontsize=14, fontweight='bold')
axes[1, 1].grid(True, alpha=0.3, axis='y')

plt.tight_layout()
plt.show()

# Print performance metrics
print("=" * 50)
print("CatBoost Model Performance Summary")
print("=" * 50)
print(f"Train R² Score: {0.9836503609955745:.4f}")
print(f"Test R² Score: {0.9641453441011794:.4f}")
print(f"Train RMSE: {0.20647833040647728:.4f}")
print(f"Test RMSE: {0.30054801394141134:.4f}")
print(f"Mean Absolute Error (Test): {np.mean(prediction_error):.4f}")
print("=" * 50)
```
