## **Checking duplicates**

## **Heatmap to check the relation**

```python
df_heat=df.select_dtypes(include=np.number)
df_heat
```

```python
sns.heatmap(df_heat.corr(),annot=True,fmt='.2f',cmap='coolwarm')
plt.show()
```

## **Basic eda to understand the data**

```python
df['sex'].value_counts()
```

```python
sns.countplot(x='sex',data=df,palette='Set2')
plt.show()
```

```python
df['smoker'].value_counts()
```

```python
sns.countplot(x='smoker',data=df,palette='Set1')
plt.show()
```

```python
sns.countplot(x='sex',data=df,hue='smoker',palette='Set1')
plt.show()
```

```python
sns.scatterplot(x='total_bill',y='tip',data=df,hue='sex',palette='Set1')
plt.show()
```

```python
df['day'].value_counts()
```

```python
sns.boxplot(x='day',y='total_bill',data=df,palette='Set2')
plt.show()
```

```python
sns.countplot(x='day',hue='sex',data=df)
plt.show()
```

```python
sns.catplot(x='day',hue='sex',col='smoker',data=df,kind='count')
plt.show()
```

```python
sns.countplot(x='time',hue='sex',data=df)
plt.show()
```

```python
sns.catplot(x='time',hue='sex',col='smoker',data=df,kind='count')
plt.show()
```
